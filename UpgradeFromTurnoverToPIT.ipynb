```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

sales_purchases = pd.read_csv("missing_tax_type_taxpayer_sample_dataset_sales_and_purchases.csv")
taxpayers_tax_types_df = pd.read_csv("missing_tax_type_taxpayer_sample_dataset_registration.csv")
# import_export = pd.read_csv("missing_tax_type_taxpayer_sample_dataset_imports_and_exports.csv")

# resident_individual if have date is personal income tax 
taxpayers_tax_types_df = taxpayers_tax_types_df[["masked_pin","gender","year_of_birth","resident_individual","turnover_tax"]]

taxpayers_tax_types_df['resident_individual'] = taxpayers_tax_types_df['resident_individual'].notna().astype(int)
taxpayers_tax_types_df['turnover_tax'] = taxpayers_tax_types_df['turnover_tax'].notna().astype(int)

taxpayers_tax_types_df = taxpayers_tax_types_df.fillna(
    {'gender': 'MISSING'}
)
```

```python
sales_purchases = sales_purchases.groupby(["masked_pin","year"]).agg({
    "b2b_nber_of_invoices":sum,	"b2b_sales_amount":sum,	"b2c_nber_of_invoices":sum,	"b2c_sales_amount":sum,	"purchases_nber_of_invoices":sum,	"purchases_amount":sum
}).reset_index()

upgrade_from_turnover_tax = pd.merge(sales_purchases, taxpayers_tax_types_df, left_on="masked_pin", right_on="masked_pin", how="left")

upgrade_from_turnover_tax["Sales"] = upgrade_from_turnover_tax["b2b_sales_amount"] + upgrade_from_turnover_tax["b2c_sales_amount"]
upgrade_from_turnover_tax["Purchases"] = upgrade_from_turnover_tax["purchases_amount"]

upgrade_from_turnover_tax = upgrade_from_turnover_tax[(upgrade_from_turnover_tax["year"] == upgrade_from_turnover_tax["year"].max()) & (upgrade_from_turnover_tax["Sales"] >= 25000000) & (upgrade_from_turnover_tax["turnover_tax"] == 1) & (upgrade_from_turnover_tax["resident_individual"] == 0)].reset_index(drop=True)
upgrade_from_turnover_tax[['masked_pin','gender','year_of_birth', 'resident_individual', 'turnover_tax', 'year', 'b2b_nber_of_invoices', 'b2b_sales_amount',
       'b2c_nber_of_invoices', 'b2c_sales_amount','purchases_nber_of_invoices', 'purchases_amount', 'Sales','Purchases']]
```
